**Цель работы:** осуществить предварительную обработку данных CSV-файла (visits.csv), выявить и устранить проблемы в этих данных.

Набор данных visits.csv содержит сведения о пользовательских сессиях в интернет-магазине.  
Каждая строка представляет собой визит пользователя со следующими признаками:
1. user_id — уникальный идентификатор пользователя; 
2. region — страна / регион пользователя;
3. device — тип устройства (iPhone, Mac, Android, PC);
4. channel — канал привлечения (organic, TipTop, FaceBoom, MediaTornado);
5. session_start — дата и время начала сессии;
6. session_end — дата и время завершения сессии;
7. time_session — длительность сессии в минутах;
8. click_count — число переходов / кликов за сессию;
9. buy_count — количество оформленных покупок;
10. price — общая стоимость покупок за сессию;
11. age — возраст пользователя.

# 1. Загрузка набора данных

```python
import pandas as pd

df = pd.read_csv('visits.csv', sep=';')
print(f"Размер датасета: {df.shape[0]} строк, {df.shape[1]} столбцов")
```

#### Проверка и нормализация наименований столбцов

```python
print("Исходные названия столбцов:", list(df.columns))
df.columns = df.columns.str.strip().str.lower()
print("Нормализованные названия столбцов:", list(df.columns))
```

## 3. Проверка и устранение пропущенных значений

```python
print("Количество пропусков до обработки:")
print(df.isna().sum())

mode_region = df['region'].mode()[0]
mode_device = df['device'].mode()[0]

df['region'] = df['region'].fillna(mode_region)
df['device'] = df['device'].fillna(mode_device)

print("\nКоличество пропусков после обработки:")
print(df.isna().sum())
print("\nИтого оставшихся пропусков:", df.isna().sum().sum())
```

## 4. Проверка и устранение дубликатов

```python
print("Количество полных явных дубликатов:", df.duplicated().sum())

print("\nУникальные значения region до унификации:")
print(df['region'].value_counts())

print("\nУникальные значения device до унификации:")
print(df['device'].value_counts())

df['region'] = df['region'].replace({'USA': 'United States'})
df['device'] = df['device'].replace({'MAC': 'Mac', 'IPHONE': 'iPhone'})

print("\nРаспределение region после унификации:")
print(df['region'].value_counts())

print("\nРаспределение device после унификации:")
print(df['device'].value_counts())
```

## 5. Проверка и преобразование типов данных

```python
df['session_start'] = pd.to_datetime(df['session_start'], format='%d.%m.%Y %H:%M')
df['session_end'] = pd.to_datetime(df['session_end'], format='%d.%m.%Y %H:%M')

print(df.dtypes)
```

## 6. Группировки и сводные таблицы 

### Задание 1. Группировка region и количество устройств (device)

```python
task1_group = df.groupby(['region', 'device'])['user_id'].count()
task1_group
```

### Задание 2. Группировка device и количество рекламных источников каждого типа (channel)

```python
task2_df = df.groupby(['device', 'channel']).size().reset_index(name='count')
task2_df = task2_df.sort_values(by='count', ascending=False).reset_index(drop=True)
task2_df
```

### Задание 3. Сводная таблица (pivot_table) — количество пользователей для каждого канала (channel)

```python
task3_pivot = df.pivot_table(index='channel', values='user_id', aggfunc='count').rename(columns={'user_id': 'count'})
task3_pivot = task3_pivot.sort_values(by='count', ascending=False)
task3_pivot
```

### Задание 4. Сводная таблица (pivot_table) — количество пользователей для каждого устройства (device — строки) и канала (channel — столбцы)

```python
task4_pivot = df.pivot_table(index='device', columns='channel', values='user_id', aggfunc='count', fill_value=0)
task4_pivot = task4_pivot.sort_index(ascending=True)
task4_pivot
```
